```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import math
```

```python
plt.figure(figsize=(20, 20))
n_cols = len(df_new.columns)
grid_size = math.ceil(math.sqrt(n_cols))

for i, col in enumerate(df_new.columns):
    plt.subplot(grid_size, grid_size, i + 1)
    sns.histplot(df_new[col], kde=True)

    plt.title(f'distribution of {col}')
plt.tight_layout()
```

```python
plt.figure(figsize=(20, 20)) # Adjusted figure size for better visualization of many plots
n_cols = len(df_new.columns) # Changed to df_new.columns for consistency with the loop
grid_size = math.ceil(math.sqrt(n_cols))

for i, col in enumerate(df_new.columns):
    plt.subplot(grid_size, grid_size, i + 1)
    sns.boxplot(df_new[col])

    plt.title(f'distribution of {col}')
plt.tight_layout()
```
